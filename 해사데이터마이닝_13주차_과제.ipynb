```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf

df = pd.read_csv('churn_logit1.csv')

print("=== 데이터 미리보기 ===")
print(df.head())

print("\n=== 기본 정보 ===")
print(df.info())

model = smf.logit('churn ~ age + usage_hour + complaint_cnt', data=df).fit()

print("\n=== 로지스틱 회귀 요약 ===")
print(model.summary())

result_table = pd.DataFrame({
    'coef': model.params,
    'p_value': model.pvalues,
    'odds_ratio': np.exp(model.params)
})

print("\n=== 계수 / p-value / 오즈비 ===")
print(result_table)

conf = model.conf_int()
conf.columns = ['2.5%', '97.5%']
conf['OR_2.5%'] = np.exp(conf['2.5%'])
conf['OR_97.5%'] = np.exp(conf['97.5%'])

print("\n=== 신뢰구간 ===")
print(conf)

df['pred_prob'] = model.predict(df)
df['pred_class'] = (df['pred_prob'] >= 0.5).astype(int)

print("\n=== 예측 결과 ===")
print(df[['churn', 'pred_prob', 'pred_class']].head())

print("\n=== 해석 가이드 ===")
for var in ['age', 'usage_hour', 'complaint_cnt']:
    coef = model.params[var]
    pval = model.pvalues[var]
    or_val = np.exp(coef)
    direction = '증가' if coef > 0 else '감소'

    print(f"{var}: 계수={coef:.4f}, p-value={pval:.6f}, 오즈비={or_val:.4f}")
    print(f" → {var}가 1 증가하면 이탈 odds는 {or_val:.4f}배 ({direction})")
```

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import accuracy_score

df = pd.read_csv('loan_default_logit2.csv')

print("=== 데이터 미리보기 ===")
print(df.head())

print("\n=== 기본 정보 ===")
print(df.info())

model = smf.logit('default ~ income + debt_ratio + late_cnt', data=df).fit()

print("\n=== 로지스틱 회귀 요약 ===")
print(model.summary())

result_table = pd.DataFrame({
    'coef': model.params,
    'p_value': model.pvalues,
    'odds_ratio': np.exp(model.params)
})

print("\n=== 계수 / p-value / 오즈비 ===")
print(result_table)

llf = model.llf
residual_deviance = -2 * llf

print("\n=== 적합도 지표 ===")
print(f"log-likelihood: {llf:.4f}")
print(f"residual deviance: {residual_deviance:.4f}")

df['pred_prob'] = model.predict(df)
df['pred_class'] = (df['pred_prob'] >= 0.5).astype(int)

# 성능
acc = accuracy_score(df['default'], df['pred_class'])
err = 1 - acc

print("\n=== 분류 성능 ===")
print(f"accuracy   : {acc:.4f}")
print(f"error rate : {err:.4f}")

print("\n=== 예측 결과 일부 ===")
print(df[['default', 'pred_prob', 'pred_class']].head(10))

conf = model.conf_int()
conf.columns = ['2.5%', '97.5%']
conf['OR_2.5%'] = np.exp(conf['2.5%'])
conf['OR_97.5%'] = np.exp(conf['97.5%'])

print("\n=== 신뢰구간 ===")
print(conf)

print("\n=== 해석 가이드 ===")
for var in ['income', 'debt_ratio', 'late_cnt']:
    coef = model.params[var]
    pval = model.pvalues[var]
    or_val = np.exp(coef)

    print(f"{var}: 계수={coef:.6f}, p-value={pval:.6f}, 오즈비={or_val:.6f}")
```
