```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import statsmodels.formula.api as smf
```

```python
file_path = 'retail_sales_reg1.csv'

df = pd.read_csv(file_path)

print('[데이터 상위 5행]')
print(df.head(), '\n')

print('[기본 정보]')
print(df.info(), '\n')

model = smf.ols('sales ~ ad_cost + staff + event_cnt', data=df).fit()

print('[회귀 요약 결과]')
print(model.summary())

print('\n[회귀계수]')
print(model.params)

print('\n[p-value]')
print(model.pvalues)

print('\n[R-squared]')
print(model.rsquared)

beta_ad = model.params['ad_cost']
print('\n광고비 10단위 증가 시 예상 매출 변화:', beta_ad * 10)

sig_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
print('\n유의한 변수:', sig_vars)
```

```python
from google.colab import files
uploaded = files.upload()

import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
from sklearn.metrics import mean_squared_error
```

```python
file_path = 'weather_reg2.csv'

df = pd.read_csv(file_path)

print('[데이터 상위 5행]')
print(df.head(), '\n')

print('[기본 정보]')
print(df.info(), '\n')

model = smf.ols('temperature ~ solar + wind + o3', data=df).fit()

print('[회귀 요약 결과]')
print(model.summary())

print('\n[회귀계수]')
print(model.params)

print('\n[p-value]')
print(model.pvalues)

print('\n[R-squared]')
print(model.rsquared)

new_df = pd.DataFrame({
    'solar': [100],
    'wind': [5],
    'o3': [30]
})

pred = model.predict(new_df)
print('\n[예측값]')
print(pred)

pred_frame = model.get_prediction(new_df).summary_frame(alpha=0.05)
print('\n[예측 요약표]')
print(pred_frame)

y_true = df['temperature']
y_pred = model.predict(df)

mse = mean_squared_error(y_true, y_pred)
rmse = np.sqrt(mse)

print('\n[MSE / RMSE]')
print('MSE :', mse)
print('RMSE:', rmse)
```
